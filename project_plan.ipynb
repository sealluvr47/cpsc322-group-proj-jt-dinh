# Brain Aneurysm Classification Using Machine Learning

**Team members** - Jason & Dinh Dinh  
**CPSC 322: Project Proposal**  


## 1. Overview

Our project investigates whether machine learning algorithms can classify brain aneurysm detection using patient demographics and imaging information. We are interested in this topic because brain aneurysms are an important medical condition, and we want to explore how machine learning can be applied to medical data.

Our main research question is: How well can machine learning algorithms classify brain aneurysm detection using patient demographics and imaging modality, and which attributes contribute most to the predictions?

We will compare different classification algorithms and test whether adding more attributes improves their performance. Our goal is to understand how well the available information can distinguish between imaging records with and without detected aneurysms.

## 2. Dataset Description

### Source 1 — RSNA Intracranial Aneurysm Detection Dataset

- **Source link:** https://imaging.rsna.org/dataset/7
- **File format:** CSV (`dataset.csv`).
- **Contents:** The dataset contains patient demographics, imaging information, and brain aneurysm detection results. Each row represents one imaging series, and some patients have multiple records.
- **Dataset size:** The dataset contains 5,041 imaging records from 3,251 unique patients, with 25 attributes.
- **Target column:** `Aneurysm Present`, which has two classes:
  - 0 = No aneurysm detected (2,888 records).
  - 1 = Aneurysm detected (2,153 records).
- **Access and reuse:** The dataset is available through the RSNA website. We will download and use the CSV file for our project without needing to download the medical images.

### Input Attributes

We plan to use five attributes for classification:

- **PatientAge:** The patient's age.
- **PatientSex:** The patient's sex.
- **Race:** The patient's recorded race.
- **Ethnicity:** The patient's recorded ethnicity.
- **Modality:** The type of imaging examination used.

We will compare different combinations of these attributes to see which ones help improve classification. We will not use patient identifiers or aneurysm location columns as predictors because they could cause data leakage.

## 3. Implementation and Technical Merit

### Preprocessing and Classification Challenges

One challenge is that our dataset contains both numerical and categorical attributes. We will need to convert categorical values, such as sex, race, ethnicity, and imaging modality, into a format that our algorithms can use.

Another challenge is that some patients have multiple imaging records. We will make sure records from the same patient stay together when splitting the data into training and testing sets to prevent data leakage.

The dataset also has more records without detected aneurysms than records with detected aneurysms. We will examine how this class imbalance affects our models and evaluate them using accuracy, precision, recall, F1-score, and confusion matrices.

### Classification Approach

We plan to compare three classification algorithms:

1. **Decision Tree:** Uses different attributes to split the data and classify whether an aneurysm was detected.
2. **Random Forest:** Combines multiple decision trees to make predictions. We plan to use our custom `MyRandomForestClassifier` from the course.
3. **Naive Bayes:** Uses probabilities based on patient attributes to predict whether an aneurysm was detected.

We will compare the algorithms using the same training and testing partitions to determine which performs best.

### Feature Selection

Since our dataset has only five input attributes, we do not need extensive feature selection. Instead, we will compare three combinations of attributes:

- **Experiment A:** Age and sex.
- **Experiment B:** Age, sex, and imaging modality.
- **Experiment C:** Age, sex, imaging modality, race, and ethnicity.

We will compare the results to see whether adding more attributes improves classification performance. We will also examine which attributes contribute most to the predictions.

We will exclude patient identifiers and aneurysm location columns because they are not appropriate predictors and could cause data leakage.

## 4. Potential Impact of the Results

Our project will help us understand how well machine learning algorithms can classify brain aneurysm detection using patient demographics and imaging information. By comparing different algorithms, we can see which one performs best and which attributes are most useful for classification.

The results may be interesting to medical researchers, radiologists, and people studying machine learning in healthcare. They could help researchers understand whether basic patient information provides useful patterns related to aneurysm detection.

However, our dataset does not include detailed medical history or the actual medical images as model inputs, so our predictions may be limited. We will also consider how differences in patient populations and imaging practices could affect the results.

Our goal is to explore the potential and limitations of machine learning using medical data, rather than develop a tool for medical diagnosis.
