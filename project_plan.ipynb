## 1. Overview

Our project investigates whether patient characteristics can help predict brain aneurysms detected through imaging. 

Our primary classification question is **“Can we predict whether a brain aneurysm was detected on a scan using patient information available before the scan result?”** The target will have two classes, aneurysm detected and no aneurysm detected.

The kidney disease component will depend on finding data that includes kidney information and confirmed aneurysm outcomes for the same individuals. We will finalize the project’s scope after reviewing the available datasets.

### Anticipated Attributes and Class Information

Depending on availability, our predictors may include age, sex, blood pressure, hypertension history, smoking history, diabetes, and family history of aneurysms. Kidney-related attributes of interest include kidney disease diagnosis, ADPKD status, kidney function measurements, dialysis history, and prior kidney transplantation. We will confirm the exact attributes after inspecting the selected data.

Our class label will indicate whether an aneurysm was detected on the selected imaging examination. Records with unknown screening results will require separate handling. We will examine aneurysm measurements, location labels, and treatment information for target leakage because these fields may reveal the outcome we are trying to predict.

If we combine datasets, we will first compare their populations, attribute definitions, units, and outcome labels. Connecting information about kidney disease and brain imaging across separate files requires a valid shared patient identifier.

## 3. Implementation and Technical Merit

### Preprocessing and Classification Challenges

We anticipate missing values, inconsistent units, categorical variables, and differences in how medical conditions are recorded. We will inspect these issues and select appropriate cleaning, encoding, and discretization methods. Any preprocessing that learns from the data will be fitted using training data and applied to evaluation data.

The dataset may contain substantially more aneurysm-negative cases than positive cases. We will examine the class distribution and evaluate our classifiers using accuracy, precision, recall, F1 score, and confusion matrices. A majority-class baseline will help us determine whether the models improve on always predicting the most common outcome.

Some patients may contribute multiple scans or aneurysms. We will identify repeated records and keep each patient within a single training or evaluation partition. We will also check for overlapping records across sources and investigate whether differences between datasets affect performance.

### Classification Approach

We plan to compare at least three classifiers from the course, including a decision tree and the required custom `MyRandomForestClassifier`. Naive Bayes is a candidate for the third classifier, depending on the selected data.

We will follow test-driven development for the random forest implementation and explore different values for the number of generated trees, retained trees, and candidate attributes at each split. We will repeat randomized evaluations and compare the forest with a standard decision tree.

A neural network may be explored as an additional experiment if the data, project timeline, and course requirements support it. Any external implementation or supporting code will be cited.

### Feature Selection

If many attributes are available, we will use exploratory data analysis to identify attributes with extensive missing values, redundant information, or potential target leakage. We will compare a smaller, interpretable feature set with a broader set using validation within the training data.

If kidney-related variables are available, we will compare classifier performance with and without those variables to assess whether they provide additional predictive information.

## 4. Potential Impact of the Results

This project could improve our understanding of which patient characteristics are associated with brain aneurysm detection. Exploring kidney disease and transplant history may identify patterns that support further research into screening for these populations.

The usefulness of our findings will depend on the quality and representativeness of the selected datasets. We will discuss how missing information, screening practices, and differences between patient populations affect our conclusions. Further clinical validation would be necessary before applying the models to patient care.

Interested stakeholders include nephrologists, neurologists, neurosurgeons, radiologists, transplant teams, primary care clinicians, and medical researchers studying aneurysms or kidney disease. Patients with kidney disease and their families may also benefit from research that improves understanding of aneurysm detection and screening.
