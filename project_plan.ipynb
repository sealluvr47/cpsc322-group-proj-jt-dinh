## 1. Overview

Our project investigates whether machine learning algorithms can classify reported diabetes status using demographic information, overall health, and lifestyle habits. 

Our main research question is **“How accurately can machine learning algorithms classifydiabetes using basic health and lifestyle information”**

We will compare decision tree, random forest, and Naive Bayes classifiers. Gradient forest would be a strech goal. 

## 2. Dataset 

### Source — Diabetes Health Indicators Dataset

- **Dataset download** — [Diabetes Health Indicators on Kaggle](https://www.kaggle.com/datasets/alexteboul/diabetes-health-indicators-dataset)
- **Repository description** — [CDC Diabetes Health Indicators on UCI](https://archive.ics.uci.edu/dataset/891/cdc+diabetes+health+indicators)
- **Original survey documentation** — [CDC BRFSS 2015](https://www.cdc.gov/brfss/annual_data/annual_2015.html)
- **Selected file** — `diabetes_binary_health_indicators_BRFSS2015.csv`
- **File format** — CSV.
- **Contents** — A cleaned selection of health and lifestyle survey responses derived from the CDC’s 2015 Behavioral Risk Factor Surveillance System. Each row represents one survey respondent.
- **Dataset size** — 253,680 responses with 21 predictor attributes and one target column.
- **Missing values** — The repository reports no missing values. We will verify this after loading the selected file.
- **Target column** — `Diabetes_binary`.
- **Target classes** — `0` represents no reported diabetes, and `1` represents reported prediabetes or diabetes.


### Selected Input Attributes

| Attribute | Description |
|---|---|
| `Age` | The respondent’s age category, encoded into 13 groups. |
| `Sex` | The respondent’s recorded sex, encoded as 0 for female and 1 for male. |
| `BMI` | Body mass index. |
| `HighBP` | Whether the respondent reports having been told they have high blood pressure. |
| `HighChol` | Whether the respondent reports having been told they have high cholesterol. |
| `GenHlth` | Self-rated general health, ranging from 1 for excellent to 5 for poor. |
| `PhysActivity` | Whether the respondent reports physical activity during the past 30 days, excluding their regular job. |
| `Smoker` | Whether the respondent has smoked at least 100 cigarettes during their lifetime. |

Perhaps we will select more as we get into the feature experimentation process. 

## 3. Implementation and Technical Merit

### Preprocessing and Classification Challenges

Most selected attributes are already numerically encoded, but their meanings differ. For example, `Age` represents an age category, and `GenHlth` represents an ordered health rating. We will relabel these as needed as we progress through the dataset. 

BMI has many possible values. We will have to normalize it before use in our clean dataset. 

Evaluation will include accuracy, precision, recall, F1 score. 

### Classification

We plan to compare three classification algorithms.

1. **Decision Tree** — Uses entropy-based attribute selection to classify reported diabetes status.
2. **Random Forest** — Uses our custom `MyRandomForestClassifier` to combine predictions from multiple decision trees.
3. **Naive Bayes** — Uses probabilities estimated from respondent attributes to classify reported diabetes status.

One issue that we see might happen is that this dataset includes far more samples than expected. We may cut down the dataset or perhaps only filter through and only include some attributes over others. 

### Feature Selection

- **Experiment A** — `Age`, `Sex`, and `BMI`.
- **Experiment B** — Experiment A plus `HighBP` and `HighChol`.
- **Experiment C** — Experiment B plus all attributes.

We have formatted it this way to later compare the experiments and see which features dicate more towards diabetes. 

## 4. Potential Impact of the Results

Helping doctors understand how demographic information, health indicators, and lifestyle habits relate to reported diabetes status. Comparing classifiers will demonstrate how different approaches handle the same medical classification problem.

Potential stakeholders include primary care doctors & medical students learning to understand basic diabetes principles.
